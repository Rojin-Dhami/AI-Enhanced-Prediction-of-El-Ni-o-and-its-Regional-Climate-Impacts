**Step 4: Data Cleaning and Type Conversion**
*   **Action:** Convert the 'time' column to a datetime object. The initial check shows a large number of missing values, especially for `sst` over land and a systematic issue in the South Asia dataset. We will use spatial interpolation for `sst` and investigate the other missing data before deciding on a filling strategy.
*   **Reason:** A simple forward fill is not appropriate for spatially-gridded data with large gaps like land masses. Spatial interpolation will provide a more physically realistic estimate. For the other variables, we need to understand the cause of the missing data before filling.


```python
# Convert 'time' column to datetime
pacific_df['time'] = pd.to_datetime(pacific_df['time'])
south_asia_df['time'] = pd.to_datetime(south_asia_df['time'])

```

**Step 5: Climfill -style gap filling for SST**

```python
# ============================================================================
# CLIMFILL-style gap filling for SST (Bessenbacher et al., 2022)
# Combines: (1) spatial interpolation -> (2) feature engineering ->
#           (3) multivariate iterative statistical imputation (refinement).
#
# IMPORTANT: SST is physically undefined over land, so the pipeline is applied
# to OCEAN points only. Ocean is identified via the land-sea mask (lsm < 0.5);
# land SST is left as NaN. NOTE: 'land_ocean' is only populated at missing-SST
# points, so it cannot be used to select the ocean training data.
# ============================================================================
from sklearn.experimental import enable_iterative_imputer  # noqa: F401
from sklearn.impute import IterativeImputer
from sklearn.linear_model import BayesianRidge
from scipy.interpolate import griddata

def climfill_sst(df):
    missing_before = df['sst'].isnull().sum()

    # --- Work on the ocean subset only (lsm < 0.5 == sea) ---
    ocean_mask = df['lsm'] < 0.5
    ocean = df.loc[ocean_mask].copy()
    ocean['month'] = ocean['time'].dt.month

    # Keep the observed SST range so we can clip the imputed values to a plausible band
    obs_min, obs_max = ocean['sst'].min(), ocean['sst'].max()

    # ----------------------------------------------------------------------------
    # Step 1 - Spatial interpolation: build a first-guess for every ocean gap by
    # interpolating SST across lat/lon within each monthly time slice.
    # ----------------------------------------------------------------------------
    sst_interp = ocean['sst'].copy()
    for _, grp in ocean.groupby('time'):
        known = grp['sst'].notna()
        gaps = ~known
        if known.sum() >= 3 and gaps.any():
            pts = grp.loc[known, ['longitude', 'latitude']].values
            vals = grp.loc[known, 'sst'].values
            tgt = grp.loc[gaps, ['longitude', 'latitude']].values
            vi = griddata(pts, vals, tgt, method='linear')
            # Points outside the convex hull come back NaN -> fall back to nearest
            nan_i = np.isnan(vi)
            if nan_i.any():
                vi[nan_i] = griddata(pts, vals, tgt[nan_i], method='nearest')
            sst_interp.loc[grp.loc[gaps].index] = vi
    ocean['sst_interp'] = sst_interp

    # ----------------------------------------------------------------------------
    # Step 2 - Feature engineering: cyclical month + spatial coords + correlated
    # variables (multivariate predictors) alongside the spatial first-guess.
    # ----------------------------------------------------------------------------
    ocean['month_sin'] = np.sin(2 * np.pi * ocean['month'] / 12)
    ocean['month_cos'] = np.cos(2 * np.pi * ocean['month'] / 12)

    candidate_predictors = ['t2m', 'msl', 'avg_iews', 'avg_inss', 'ttr']
    predictors = [c for c in candidate_predictors if c in ocean.columns]
    feature_cols = ['latitude', 'longitude', 'month_sin', 'month_cos',
                    'sst_interp'] + predictors + ['sst']
    impute_frame = ocean[feature_cols].copy()

    # Drop any predictor columns that are entirely NaN in the ocean subset:
    # they carry no information and trigger sklearn's empty-feature code path.
    non_empty_cols = [c for c in feature_cols if impute_frame[c].notna().any()]
    impute_frame = impute_frame[non_empty_cols]

    # ----------------------------------------------------------------------------
    # Step 3 - Iterative multivariate imputation: BayesianRidge regression that
    # iteratively refines SST from the other variables until convergence.
    # Observed SST values are preserved; only missing entries are estimated.
    # ----------------------------------------------------------------------------
    imputer = IterativeImputer(estimator=BayesianRidge(), max_iter=10,
                            imputation_order='ascending', random_state=0)
    # Pass a writable float array (copy=True) to avoid the read-only assignment
    # error that Copy-on-Write can otherwise trigger inside the imputer.
    imputed = imputer.fit_transform(impute_frame.to_numpy(dtype=float, copy=True))
    sst_filled = imputed[:, non_empty_cols.index('sst')]

    # Clip to the observed SST range so refined values never look physically odd
    sst_filled = np.clip(sst_filled, obs_min, obs_max)

    # Write the filled ocean SST back into the main dataframe
    df.loc[ocean_mask, 'sst'] = sst_filled

    missing_after = df['sst'].isnull().sum()
    ocean_missing_after = df.loc[ocean_mask, 'sst'].isnull().sum()

    print(f"Missing SST before CLIMFILL:        {missing_before}")
    print(f"Missing SST after CLIMFILL:         {missing_after}")
    print(f"Land points left as NaN (expected): {(df['lsm'] >= 0.5).sum()}")

    print(f"Ocean gaps remaining:               {ocean_missing_after}")
    print(f"Land points left as NaN (expected): {(df['lsm'] >= 0.5).sum()}")
```

```python
#filling in pacific ocean sst gaps
climfill_sst(pacific_df)
```

```python
# Sanity check: the imputed ocean SST should match the observed distribution
# (no odd out-of-range values). Compare summary statistics of ocean SST.
ocean_sst = pacific_df.loc[pacific_df['lsm'] < 0.5, 'sst']
print("Ocean SST summary after CLIMFILL:")
print(ocean_sst.describe())
print(f"\nRemaining NaNs in ocean SST: {ocean_sst.isnull().sum()}")
```
