## Feature Engineering — Spatio-temporal input tensor

Construct 12-month input sequences of 4 spatial anomaly fields over the tropical Pacific
(30°S–30°N, 120°E–80°W) to forecast the Niño 3.4 ONI at a chosen lead time.

**Channels (C = 4):**
1. `sst_anom` — Sea Surface Temperature anomaly (ERA5) — primary ENSO surface indicator
2. `msl_anom` — Sea Level Pressure anomaly (ERA5) — Walker circulation / trade winds
3. `avg_iews_anom` — Zonal (east–west) wind stress anomaly (ERA5) — trade-wind driver
4. `sohtc300_anom` — Upper-300 m Ocean Heat Content anomaly (ORAS5) — subsurface predictor

Real grid is 2°×2° → **31 lat × 81 lon**, so each sequence tensor is `(12, 31, 81, 4)`.


```python
# --- Config -------------------------------------------------------------
# Four input channels (anomaly fields). We use raw anomalies (_anom), NOT the
# pre-standardised *_anom_z, because those were z-scored over the full record
# (data leakage). We standardise per-channel later, fitting on the train split only.
FEATURES = ['sst_anom', 'msl_anom', 'avg_iews_anom', 'sohtc300_anom']

SEQ_LEN = 12      # months per input window
LEAD    = 3       # forecast horizon: predict ONI this many months AFTER window end
TARGET_COL = 'nino34_3m'   # 3-month running mean Niño 3.4 = official ONI

# Canonical, sorted axes (grid is regular & complete)
times = np.sort(pacific_df['time'].unique())
lats  = np.sort(pacific_df['latitude'].unique())
lons  = np.sort(pacific_df['longitude'].unique())
T, H, W, C = len(times), len(lats), len(lons), len(FEATURES)

print(f"T={T} months | H={H} lats | W={W} lons | C={C} channels")
print(f"Sequence length={SEQ_LEN}, lead={LEAD} months, target='{TARGET_COL}'")
```

```python
# --- Build the full spatio-temporal tensor: (T, H, W, C) -----------------
# Reindex onto the complete (time x lat x lon) grid so any missing cell -> NaN,
# guaranteeing correct ordering regardless of the source row order.
full_idx = pd.MultiIndex.from_product([times, lats, lons],
                                      names=['time', 'latitude', 'longitude'])
gridded = (pacific_df
           .set_index(['time', 'latitude', 'longitude'])[FEATURES]
           .reindex(full_idx))

grid = np.full((T, H, W, C), np.nan, dtype=np.float32)
for c, feat in enumerate(FEATURES):
    grid[..., c] = gridded[feat].to_numpy(dtype=np.float32).reshape(T, H, W)

# Land cells are NaN for ocean-only fields (SST, OHC). Record a land mask,
# then fill NaNs with 0.0 (neutral anomaly) so models receive finite values.
nan_fraction = np.isnan(grid).mean(axis=(0, 1, 2))
grid = np.nan_to_num(grid, nan=0.0)

print("Full grid tensor shape (T, H, W, C):", grid.shape)
for feat, frac in zip(FEATURES, nan_fraction):
    print(f"  {feat:<16} NaN filled: {frac*100:5.1f}%")
```

```python
# --- Align ONI target on the same monthly time axis ----------------------
# Map each timestep -> ONI value (nino34_3m). reindex(times) guarantees the
# target array is in exactly the same order as the tensor's time axis.
target_series = nino.set_index('time')[TARGET_COL]
target_arr = target_series.reindex(times).to_numpy(dtype=np.float32)

print("Target aligned:", target_arr.shape)
print("NaNs in target:", np.isnan(target_arr).sum(),
      "(expect a few at the very start: 3-month running mean warm-up)")
```

```python
# --- Build sliding-window samples ----------------------------------------
# Window i covers months [i, i+SEQ_LEN-1]; the target is the ONI at month
# (i + SEQ_LEN - 1 + LEAD), i.e. LEAD months after the window's last month.
X_list, y_list, sample_times = [], [], []

last_start = T - SEQ_LEN - LEAD + 1        # exclusive upper bound for i
for i in range(last_start):
    t_idx = i + SEQ_LEN - 1 + LEAD          # index of the forecast month
    y_val = target_arr[t_idx]
    if np.isnan(y_val):                     # skip warm-up NaNs
        continue
    X_list.append(grid[i:i + SEQ_LEN])      # (SEQ_LEN, H, W, C)
    y_list.append(y_val)
    sample_times.append(times[t_idx])       # the month being predicted

X = np.stack(X_list).astype(np.float32)     # (N, SEQ_LEN, H, W, C)
y = np.asarray(y_list, dtype=np.float32)    # (N,)
sample_times = np.asarray(sample_times)

print("X shape:", X.shape, "  y shape:", y.shape)
print("First predicted month:", sample_times[0], "| last:", sample_times[-1])
```

```python
# --- Sanity checks --------------------------------------------------------
# 1) No NaN/Inf leaked into the inputs or targets
assert np.isfinite(X).all(), "X contains NaN/Inf"
assert np.isfinite(y).all(), "y contains NaN/Inf"

# 2) Temporal alignment: for sample k, target month == window_last_month + LEAD
k = 100
win_last_month = times[np.where(times == sample_times[k])[0][0] - LEAD]
print(f"Sample {k}: window ends {win_last_month} -> predicts {sample_times[k]} "
      f"(+{LEAD} mo)  |  y = {y[k]:+.3f}")

# 3) Flattened length for the XGBoost baseline
flat_len = SEQ_LEN * H * W * C
print(f"\nFlattened XGBoost vector length = {SEQ_LEN}*{H}*{W}*{C} = {flat_len:,}")

# 4) Target distribution
print(f"\nONI target  min={y.min():+.2f}  max={y.max():+.2f}  "
      f"mean={y.mean():+.2f}  std={y.std():.2f}")
print("Samples:", X.shape[0], "| tensor per sample:", X.shape[1:])
```

## Feature Selection — MIFS (Mutual Information, redundancy-penalized)

Applied **only to the XGBoost baseline** (the CNN-TCN uses the full tensor).
Pipeline, all fit on the **training split only** to avoid leakage:

1. **Chronological split** — train ≤ 2018, validation 2019–2022, test 2023–2026.
2. **Standardize** each channel using train-set mean/std, then flatten to `(N, 120528)`.
3. **Relevance** — mutual information `I(X_ij ; Y)` between every grid-cell feature and
   the continuous ONI target (`mutual_info_regression`); constant land cells dropped.
4. **MIFS greedy selection** — iteratively pick the feature maximizing
   `S = I(X_ij;Y) − (1/|S|) Σ I(X_ij;X_kl)` (relevance minus mean redundancy to the
   already-selected set). Redundancy uses the Gaussian-MI approximation
   `I ≈ −½ ln(1−ρ²)` for tractability over thousands of candidates.
5. **Choose K** — the number of retained features that minimizes **validation RMSE**
   of an XGBoost model.


```python
# --- Chronological train / val / test split ------------------------------
# Split on the predicted month (sample_times). No shuffling: this is a time series.
sample_dt = pd.to_datetime(sample_times)

train_mask = sample_dt <= pd.Timestamp('2018-12-31')
val_mask   = (sample_dt >= pd.Timestamp('2019-01-01')) & (sample_dt <= pd.Timestamp('2022-12-31'))
test_mask  = sample_dt >= pd.Timestamp('2023-01-01')

for name, m in [('train', train_mask), ('val', val_mask), ('test', test_mask)]:
    dts = sample_dt[m]
    print(f"{name:5s}: n={m.sum():4d}  {dts.min().date()} -> {dts.max().date()}")

assert (train_mask.sum() + val_mask.sum() + test_mask.sum()) == len(sample_dt)
```

```python
# --- Standardize per channel (train stats only) then flatten -------------
# Per-channel mean/std computed over TRAIN samples across (sample, time, lat, lon).
mu = X[train_mask].mean(axis=(0, 1, 2, 3))          # (C,)
sd = X[train_mask].std(axis=(0, 1, 2, 3)) + 1e-8    # (C,)
X_std = (X - mu) / sd

# Flatten each sample: (12, 31, 81, 4) -> 120,528
X_flat = X_std.reshape(X.shape[0], -1).astype(np.float32)

X_tr, X_va, X_te = X_flat[train_mask], X_flat[val_mask], X_flat[test_mask]
y_tr, y_va, y_te = y[train_mask],      y[val_mask],      y[test_mask]

print("Per-channel mean:", np.round(mu, 4))
print("Per-channel std :", np.round(sd, 4))
print("X_tr:", X_tr.shape, "| X_va:", X_va.shape, "| X_te:", X_te.shape)
```

```python
# --- Relevance: mutual information I(X_ij ; Y) on the training split ------
# Drop constant features (land cells for SST/OHC) — they carry zero information.
from sklearn.feature_selection import mutual_info_regression
import time

feat_std = X_tr.std(axis=0)
active = np.where(feat_std > 1e-8)[0]          # flat indices of informative cells
print(f"Active (non-constant) features: {active.size:,} / {X_tr.shape[1]:,}")

t0 = time.time()
mi = mutual_info_regression(X_tr[:, active], y_tr, random_state=0)
print(f"Computed relevance MI for {active.size:,} features in {time.time()-t0:.1f}s")
print(f"MI range: min={mi.min():.4f}  max={mi.max():.4f}  mean={mi.mean():.4f}")
```

```python
# --- Greedy MIFS with redundancy penalization ----------------------------
# Redundancy over 120k features is intractable pairwise, so restrict the greedy
# search to the top-P most relevant candidates, then penalize redundancy within
# that pool. Redundancy MI uses the Gaussian approximation I ≈ -0.5 ln(1 - r^2).
P     = 3000    # candidate pool size (top-P by relevance)
K_MAX = 800     # max features to rank via greedy MIFS

order = np.argsort(mi)[::-1][:P]         # positions into `active`
cand  = active[order]                    # global flat indices of candidates
rel   = mi[order].astype(np.float64)     # relevance I(X;Y) for candidates

# Pairwise redundancy MI among candidates (train data)
Xc = X_tr[:, cand]
corr = np.corrcoef(Xc, rowvar=False)
corr = np.clip(corr, -0.999999, 0.999999)
red_mi = -0.5 * np.log1p(-corr**2)       # (P, P) redundancy MI approximation
np.fill_diagonal(red_mi, 0.0)

# Greedy selection: start with the most relevant feature, then maximize
# S = relevance - mean redundancy to already-selected set.
selected = [int(np.argmax(rel))]
red_sum  = red_mi[:, selected[0]].copy()  # running sum of redundancy to selected
chosen   = np.zeros(P, dtype=bool); chosen[selected[0]] = True

while len(selected) < K_MAX:
    score = rel - red_sum / len(selected)
    score[chosen] = -np.inf
    nxt = int(np.argmax(score))
    selected.append(nxt)
    chosen[nxt] = True
    red_sum += red_mi[:, nxt]

ranked_global = cand[selected]           # flat indices, best -> worst (MIFS order)
print(f"Ranked {len(ranked_global)} features via greedy MIFS from top-{P} pool.")
print("First 5 relevance:", np.round(rel[selected[:5]], 3))
```

```python
# --- Choose K: minimize XGBoost validation RMSE --------------------------
from xgboost import XGBRegressor
from sklearn.metrics import mean_squared_error

def make_xgb():
    return XGBRegressor(
        n_estimators=400, max_depth=4, learning_rate=0.05,
        subsample=0.8, colsample_bytree=0.8,
        reg_lambda=1.0, random_state=0, n_jobs=-1,
    )

K_grid = [10, 25, 50, 75, 100, 150, 200, 300, 400, 500, 600, 700, 800]
val_rmse = []
for K in K_grid:
    idx = ranked_global[:K]
    m = make_xgb().fit(X_tr[:, idx], y_tr)
    rmse = np.sqrt(mean_squared_error(y_va, m.predict(X_va[:, idx])))
    val_rmse.append(rmse)
    print(f"K={K:4d}  val RMSE={rmse:.4f}")

best_K = K_grid[int(np.argmin(val_rmse))]
print(f"\nBest K = {best_K}  (val RMSE={min(val_rmse):.4f})")

plt.figure(figsize=(7, 4))
plt.plot(K_grid, val_rmse, 'o-')
plt.axvline(best_K, color='r', ls='--', label=f'best K={best_K}')
plt.xlabel('K (number of MIFS features)'); plt.ylabel('Validation RMSE')
plt.title('MIFS feature-count selection'); plt.legend(); plt.tight_layout()
plt.show()
```

```python
# --- Finalize the selected feature set -----------------------------------
selected_idx = ranked_global[:best_K]                 # flat indices to keep

# Reduced matrices fed to the XGBoost baseline
X_tr_sel = X_tr[:, selected_idx]
X_va_sel = X_va[:, selected_idx]
X_te_sel = X_te[:, selected_idx]
print("Selected feature matrices:",
      X_tr_sel.shape, X_va_sel.shape, X_te_sel.shape)

# Decode each flat index -> (month-lag, latitude, longitude, channel)
lag, ilat, ilon, ich = np.unravel_index(selected_idx, (SEQ_LEN, H, W, C))
sel = pd.DataFrame({
    'channel':   [FEATURES[c] for c in ich],
    'lag_month': lag,                 # 0 = oldest month in window, 11 = most recent
    'latitude':  lats[ilat],
    'longitude': lons[ilon],
})

print("\nSelected features per channel:")
print(sel['channel'].value_counts().to_string())
print("\nSelected features per month-lag (11 = most recent):")
print(sel['lag_month'].value_counts().sort_index().to_string())
```

```python
# --- Where are the selected features? (physical sanity check) ------------
fig, axes = plt.subplots(2, 2, figsize=(13, 7), sharex=True, sharey=True)
nino34_box = dict(x=190, width=240-190, y=-5, height=10)   # 5S-5N, 170W-120W

for ax, feat in zip(axes.ravel(), FEATURES):
    sub = sel[sel['channel'] == feat]
    # count selections per grid cell (features can repeat a cell across lags)
    grid_counts = sub.groupby(['longitude', 'latitude']).size().reset_index(name='n')
    sc = ax.scatter(grid_counts['longitude'], grid_counts['latitude'],
                    c=grid_counts['n'], cmap='viridis', s=40, marker='s')
    ax.add_patch(plt.Rectangle((nino34_box['x'], nino34_box['y']),
                               nino34_box['width'], nino34_box['height'],
                               fill=False, edgecolor='red', lw=1.5))
    ax.set_title(f"{feat}  (n={len(sub)})")
    ax.set_xlim(120, 280); ax.set_ylim(-30, 30)
    plt.colorbar(sc, ax=ax, label='# selected')

fig.suptitle('MIFS-selected grid cells per channel (red box = Niño 3.4)', y=1.02)
fig.supxlabel('Longitude (°E)'); fig.supylabel('Latitude (°)')
plt.tight_layout(); plt.show()
```

## Baseline Model — XGBoost on MIFS-selected features

Train the final gradient-boosting baseline on the `K = best_K` MIFS features and
evaluate it as the benchmark the CNN-TCN must beat. Reported metrics:

- **RMSE** — average error magnitude (°C of ONI)
- **MAE** — mean absolute error
- **Pearson r** — how well predicted timing/shape tracks the true ONI curve
- **R²** — fraction of ONI variance explained

Test period (2023–2026) is fully out-of-sample; train metrics are in-sample (optimistic).


```python
# --- Train the final XGBoost baseline and score it -----------------------
from scipy.stats import pearsonr

baseline = make_xgb().fit(X_tr_sel, y_tr)

def report(name, y_true, y_pred):
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    mae  = np.mean(np.abs(y_true - y_pred))
    r    = pearsonr(y_true, y_pred)[0]
    r2   = 1 - np.sum((y_true - y_pred)**2) / np.sum((y_true - y_true.mean())**2)
    print(f"{name:5s}  RMSE={rmse:.4f}  MAE={mae:.4f}  Pearson r={r:.4f}  R²={r2:.4f}")
    return dict(rmse=rmse, mae=mae, r=r, r2=r2)

pred_tr = baseline.predict(X_tr_sel)
pred_va = baseline.predict(X_va_sel)
pred_te = baseline.predict(X_te_sel)

print(f"Baseline XGBoost | K={best_K} features | lead={LEAD} months\n")
m_tr = report('train', y_tr, pred_tr)
m_va = report('val',   y_va, pred_va)
m_te = report('test',  y_te, pred_te)

baseline_metrics = {'train': m_tr, 'val': m_va, 'test': m_te,
                    'K': best_K, 'lead': LEAD}
```

```python
# --- Actual vs predicted ONI over the full timeline ----------------------
# Predictions for every sample (train region is in-sample; val/test are honest).
pred_full = baseline.predict(X_flat[:, selected_idx])

fig, ax = plt.subplots(figsize=(14, 5))
ax.plot(sample_dt, y, 'k-', lw=1.6, label='Actual ONI')
ax.plot(sample_dt, pred_full, color='C1', lw=1.2, label='XGBoost predicted')

# El Niño / La Niña thresholds
ax.axhline(0.5,  color='r', ls=':', lw=0.8)
ax.axhline(-0.5, color='b', ls=':', lw=0.8)

# Shade validation and test periods
ax.axvspan(pd.Timestamp('2019-01-01'), pd.Timestamp('2022-12-31'),
           color='orange', alpha=0.10, label='validation')
ax.axvspan(pd.Timestamp('2023-01-01'), sample_dt.max(),
           color='green', alpha=0.10, label='test')

ax.set_title(f'Niño 3.4 ONI — Actual vs XGBoost baseline '
             f'(K={best_K}, lead={LEAD} mo)')
ax.set_xlabel('Predicted month'); ax.set_ylabel('ONI (°C)')
ax.legend(loc='upper left', ncol=2); plt.tight_layout(); plt.show()
```

```python
# --- Zoom on the out-of-sample period (validation + test) ----------------
oos = sample_dt >= pd.Timestamp('2019-01-01')

fig, ax = plt.subplots(figsize=(13, 5))
ax.plot(sample_dt[oos], y[oos], 'k-o', ms=3, lw=1.6, label='Actual ONI')
ax.plot(sample_dt[oos], pred_full[oos], color='C1', marker='o', ms=3,
        lw=1.4, label='XGBoost predicted')
ax.axhline(0.5,  color='r', ls=':', lw=0.8)
ax.axhline(-0.5, color='b', ls=':', lw=0.8)
ax.axvspan(pd.Timestamp('2019-01-01'), pd.Timestamp('2022-12-31'),
           color='orange', alpha=0.10, label='validation')
ax.axvspan(pd.Timestamp('2023-01-01'), sample_dt.max(),
           color='green', alpha=0.10, label='test')
ax.set_title(f"Out-of-sample forecast (2019–2026)  |  "
             f"test RMSE={m_te['rmse']:.3f}, r={m_te['r']:.3f}")
ax.set_xlabel('Predicted month'); ax.set_ylabel('ONI (°C)')
ax.legend(loc='upper left', ncol=2); plt.tight_layout(); plt.show()
```
